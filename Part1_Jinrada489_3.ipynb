```python
# โหลดข้อมูล ============
import glob
from datetime import timezone

REPO_ID = "alpindale/two-million-bluesky-posts"
FIELDS  = ["text", "created_at", "author", "uri", "has_images", "reply_to", "predicted_language"]
N_FILES = 2

def iter_jsonl(path):
    """อ่าน JSON Lines ทีละบรรทัด ไม่โหลดทั้งไฟล์เข้า RAM"""
    with open(path, encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            try:
                yield json.loads(line)
            except json.JSONDecodeError:
                continue

def find_files(n=N_FILES):
    files = sorted(glob.glob(os.path.join(DATA_DIR, "**", "*.jsonl"), recursive=True))
    if len(files) >= n:
        return files[:n]
    # ถ้ายังไม่มีไฟล์ในเครื่อง ลองโหลดจาก Hugging Face
    from huggingface_hub import list_repo_files, hf_hub_download
    remote = sorted(f for f in list_repo_files(REPO_ID, repo_type="dataset") if f.endswith(".jsonl"))
    return [hf_hub_download(REPO_ID, f, repo_type="dataset", local_dir=DATA_DIR) for f in remote[:n]]

def norm_lang(v):
    if isinstance(v, list):
        v = v[0] if v else None
    return v if isinstance(v, str) and v else "und"

files = find_files()
print("ไฟล์ที่ใช้:", [os.path.basename(f) for f in files])

rows, per_file = [], {}
for fp in files:
    k = 0
    for r in iter_jsonl(fp):
        rows.append({
            "text": r.get("text") or "",
            "created_at": r.get("created_at"),
            "author": r.get("author"),
            "uri": r.get("uri"),
            "has_images": r.get("has_images"),
            "reply_to": r.get("reply_to"),
            "lang": norm_lang(r.get("predicted_language")),
            "file": os.path.basename(fp),
        })
        k += 1
    per_file[os.path.basename(fp)] = k

df = pd.DataFrame(rows)
del rows

# แปลงเวลา (UTC) และตรวจคุณภาพวันที่
df["ts"] = pd.to_datetime(df["created_at"], utc=True, errors="coerce")
n_raw = len(df)
n_bad_parse = int(df["ts"].isna().sum())
lo, hi = pd.Timestamp("2023-01-01", tz="UTC"), pd.Timestamp.now(tz="UTC")   # Bluesky เปิดใช้งานช่วงปี 2023
bad_range = df["ts"].notna() & ((df["ts"] < lo) | (df["ts"] > hi))
n_bad_range = int(bad_range.sum())
df = df[df["ts"].notna() & ~bad_range].copy()
df["hour_utc"] = df["ts"].dt.hour
df["hour_ict"] = df["ts"].dt.tz_convert("Asia/Bangkok").dt.hour

print(f"โหลดมา {n_raw:,} โพสต์ | ตัดวันที่อ่านไม่ได้ {n_bad_parse:,} + วันที่ผิดปกติ {n_bad_range:,} | เหลือใช้ {len(df):,}")
df.head(3)
```

# Q1) คนโพสต์เมื่อไร

```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. คำนวณสถิติพื้นฐานของเวลาและจำนวนโพสต์
N = len(df)
t_min, t_max = df["ts"].min(), df["ts"].max()

# สร้างคอลัมน์เวลาไทย (UTC+7) และนับจำนวนโพสต์แยกตามชั่วโมง
df["hour_ict"] = (df["ts"].dt.hour + 7) % 24
hourly_counts = df["hour_ict"].value_counts().reindex(range(24), fill_value=0)
hourly_pct = (hourly_counts / N) * 100

# 2. แสดงผลสรุปคำถามที่ 1
print(f"จำนวนโพสต์รวมทั้งหมด: {N:,} โพสต์")
print(
    f"ช่วงเวลาที่ครอบคลุม: {t_min:%d/%m/%Y %H:%M} ถึง {t_max:%d/%m/%Y %H:%M} UTC"
)

top3 = hourly_counts.nlargest(3)
print("\n ช่วงเวลาที่มีการโพสต์มากที่สุด 3 อันดับ (เวลาไทย):")
for hour, count in top3.items():
  print(f"   • {hour:02d}:00 - {hour:02d}:59 น. : {count:,} โพสต์ ({count/N*100:.1f}%)")

print(
    f"\n3 อันดับแรกคิดเป็น {top3.sum()/N*100:.1f}% ของโพสต์ทั้งหมดในหนึ่งวัน"
)
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# รีเซ็ตการตั้งค่าฟอนต์กลับเป็นค่า Default ของระบบ (ภาษาอังกฤษ)
plt.rcdefaults()
plt.rcParams['axes.unicode_minus'] = False

# 1. แปลงเวลาเป็น ICT (UTC+7)
df['ts'] = pd.to_datetime(df['created_at'], utc=True)
df['ts_ict'] = df['ts'].dt.tz_convert('Asia/Bangkok')
df['hour_ict'] = df['ts_ict'].dt.hour

N = len(df)
hourly_counts = df['hour_ict'].value_counts().reindex(range(24), fill_value=0)
top3_hours = hourly_counts.nlargest(3).index.tolist()

# 2. วาดกราฟภาษาอังกฤษ
plt.figure(figsize=(12, 5))

bar_colors = [
    '#ff7f0e' if h in top3_hours and hourly_counts[h] > 0 else '#1f77b4'
    for h in range(24)
]

plt.bar(
    hourly_counts.index,
    hourly_counts.values,
    color=bar_colors,
    edgecolor='black',
    alpha=0.8,
)

avg_posts = N / 24
plt.axhline(
    avg_posts,
    color='red',
    linestyle='--',
    linewidth=1.5,
    label=f'Average ({avg_posts:,.0f} posts/hr)',
)

plt.title(
    'Bluesky Posts Distribution by Hour of Day (ICT / UTC+7)',
    fontsize=13,
    fontweight='bold',
    pad=15,
)
plt.xlabel('Hour of Day (ICT)', fontsize=11)
plt.ylabel('Number of Posts', fontsize=11)
plt.xticks(range(0, 24), [f'{h:02d}:00' for h in range(24)], rotation=45)
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.legend(loc='upper left')

for h in top3_hours:
  val = hourly_counts[h]
  if val > 0:
    plt.text(
        h,
        val + (N * 0.01),
        f'{val:,}',
        ha='center',
        va='bottom',
        fontsize=9,
        fontweight='bold',
        color='#d62728',
    )

plt.tight_layout()
plt.show()
```

```python
# ============ Q2) ใช้ภาษาอะไร ============
LANG_NAME = {"en":"English","ja":"Japanese","pt":"Portuguese","es":"Spanish","de":"German","fr":"French",
             "ko":"Korean","th":"Thai","ru":"Russian","it":"Italian","tr":"Turkish","id":"Indonesian",
             "nl":"Dutch","pl":"Polish","zh":"Chinese","ar":"Arabic","und":"Undetermined"}
lname = lambda c: f"{LANG_NAME.get(c, c)} ({c})"

lang_cnt = df["lang"].value_counts()
lang_tbl = pd.DataFrame({"ภาษา": [lname(c) for c in lang_cnt.index],
                         "โพสต์": lang_cnt.values,
                         "สัดส่วน %": (lang_cnt.values / N * 100).round(2)})
lang_tbl["สะสม %"] = lang_tbl["สัดส่วน %"].cumsum().round(2)
display(lang_tbl.head(10))

known = lang_cnt.drop("und", errors="ignore")

# ตรวจสอบว่ามีข้อมูลภาษาที่จำแนกได้หรือไม่
if known.empty:
    print("ไม่พบข้อมูลภาษาที่จำแนกได้ (ข้อมูลมีเฉพาะ 'und' หรือตารางข้อมูลว่างเปล่า)")
    print(f"ภาษาที่ระบบจำแนกไม่ได้ (und): {lang_cnt.get('und',0)/N*100:.1f}% | ภาษาไทย (th): {lang_cnt.get('th',0)/N*100:.2f}% ({lang_cnt.get('th',0):,} โพสต์)")
else:
    top_codes = list(known.index[:3])
    cum = known.cumsum() / N * 100

    # คำนวณ n90 อย่างปลอดภัย
    n90 = int((cum < 90).sum() + 1) if (len(cum) > 0 and cum.iloc[-1] >= 90) else len(cum)

    # ดึงรายชื่อภาษาตามจำนวนที่มีจริง (สูงสุด 3 อันดับแรก)
    top_info = []
    for i in range(len(top_codes)):
        label = "ภาษาเด่นสุด" if i == 0 else f"อันดับ {i+1}"
        top_info.append(f"{label}: {lname(top_codes[i])} {known.iloc[i]/N*100:.1f}%")

    print(" | ".join(top_info))
    print(f"ภาษาที่ระบบจำแนกไม่ได้ (und): {lang_cnt.get('und',0)/N*100:.1f}% | ภาษาไทย (th): {lang_cnt.get('th',0)/N*100:.2f}% ({lang_cnt.get('th',0):,} โพสต์)")

    top3_cov = cum.iloc[min(2, len(cum)-1)]
    print(f"ต้องใช้ {n90} ภาษา ถึงจะครอบคลุมโพสต์ >= 90% (ภาษาหลักที่พบครอบคลุม {top3_cov:.1f}%)")

    print("\n>> ข้อเสนอแนะเชิงกลยุทธ์")
    print(f"   1) ทำคอนเทนต์หลักเป็น {LANG_NAME.get(top_codes[0], top_codes[0])} ก่อน เพราะเป็นฐานใหญ่สุด ({known.iloc[0]/N*100:.1f}%)")

    if len(top_codes) > 1:
        other_langs = " และ ".join([LANG_NAME.get(c, c) for c in top_codes[1:]])
        print(f"   2) ขยายเป็น {other_langs} เป็นลำดับถัดไป รวมภาษาหลักครอบคลุม {top3_cov:.1f}%")

    th_count = lang_cnt.get('th', 0)
    ratio_diff = known.iloc[0] / max(th_count, 1)
    print(f"   3) ถ้าเป้าหมายเป็นผู้ใช้ไทย ต้องทำคอนเทนต์ภาษาไทยแยกเอง เพราะสัดส่วนไทยบนแพลตฟอร์มนี้มีเพียง {th_count/N*100:.2f}% คือน้อยกว่าภาษาหลักประมาณ {ratio_diff:.0f} เท่า")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer, ENGLISH_STOP_WORDS

# --- 3.1 Hashtag (ใช้ข้อมูลเต็ม เพราะนับเร็ว) ---
HASHTAG_RE = re.compile(r"(?<!\w)#([^\W\d_][\w]*)", re.UNICODE)
tag_counter = Counter(t.lower() for txt in df["text"] for t in HASHTAG_RE.findall(txt))
n_with_tag = int(df["text"].str.contains(r"(?<!\w)#[^\W\d_]", regex=True).sum())
print(f"โพสต์ที่มี hashtag: {n_with_tag:,} จาก {N:,} ({n_with_tag/N*100:.1f}%) | hashtag ไม่ซ้ำ {len(tag_counter):,} แท็ก")
tag_tbl = pd.DataFrame(tag_counter.most_common(15), columns=["hashtag", "จำนวนครั้ง"])
tag_tbl["% ของโพสต์ทั้งหมด"] = (tag_tbl["จำนวนครั้ง"] / N * 100).round(2)
display(tag_tbl)

# --- 3.2 คำ/วลีเด่นด้วย TF-IDF (สุ่ม SAMPLE_N โพสต์ภาษาอังกฤษ ด้วย RANDOM_SEED) ---
en = df[df["lang"] == "en"]
use = en if len(en) >= 200 else df          # กันกรณีข้อมูลภาษาอังกฤษน้อยเกินไป
samp = use.sample(n=min(SAMPLE_N, len(use)), random_state=RANDOM_SEED)

def clean(t):
    t = re.sub(r"https?://\S+", " ", t)      # ลบลิงก์
    t = re.sub(r"@[\w.\-]+", " ", t)         # ลบ mention
    return t.replace("#", " ").lower()

EXTRA_STOP = {"just","like","im","dont","ive","its","got","get","really","amp","com","www","https","http","will","one","would","also"}
vec = TfidfVectorizer(preprocessor=clean, stop_words=list(ENGLISH_STOP_WORDS | EXTRA_STOP),
                      ngram_range=(1, 2), min_df=5, max_features=30000,
                      token_pattern=r"(?u)\b[a-zA-Z][a-zA-Z']+\b")
X = vec.fit_transform(samp["text"])
terms = np.array(vec.get_feature_names_out())
doc_freq = np.asarray((X > 0).sum(axis=0)).ravel()
mean_tfidf = np.asarray(X.mean(axis=0)).ravel()
tfidf_tbl = (pd.DataFrame({"คำ/วลี": terms, "TF-IDF เฉลี่ย": mean_tfidf.round(5),
                           "จำนวนโพสต์ที่พบ": doc_freq,
                           "% ของโพสต์ที่สุ่ม": (doc_freq / X.shape[0] * 100).round(2)})
             .sort_values("TF-IDF เฉลี่ย", ascending=False))
print(f"\nTF-IDF บนโพสต์ที่สุ่ม n = {X.shape[0]:,} ({'ภาษาอังกฤษ' if use is en else 'ทุกภาษา'}), vocabulary = {len(terms):,}")
print("\nคำเดี่ยวเด่น 15 อันดับ:");  display(tfidf_tbl[~tfidf_tbl["คำ/วลี"].str.contains(" ")].head(15).reset_index(drop=True))
print("วลี 2 คำเด่น 15 อันดับ:");   display(tfidf_tbl[tfidf_tbl["คำ/วลี"].str.contains(" ")].head(15).reset_index(drop=True))

# --- 3.3 ตัวอย่างโพสต์จริง (ซ่อน author เพื่อความเป็นส่วนตัว) ---
keys = ["#" + t for t in tag_tbl["hashtag"].head(3)] + list(tfidf_tbl[tfidf_tbl["คำ/วลี"].str.contains(" ")]["คำ/วลี"].head(2))
print("\n=== ตัวอย่างโพสต์จริง ===")
for i, k in enumerate(keys, 1):
    hit = use[use["text"].str.contains(re.escape(k), case=False, regex=True) & use["text"].str.len().between(30, 300)]
    if hit.empty:
        continue
    r = hit.sample(1, random_state=RANDOM_SEED + i).iloc[0]
    print(f"\n[{i}] คำค้น: {k}  | เวลา: {r['ts']:%Y-%m-%d %H:%M} UTC | ภาษา: {r['lang']}")
    print("    " + re.sub(r"\s+", " ", r["text"]).strip()[:300])
```

### **หลักฐาน:**
จากข้อมูล n = 96,500 โพสต์ ช่วง 15:00–15:59 น. มี 78,980 โพสต์ (81.8%) และช่วง 14:00–14:59 น. มี 16,439 โพสต์ (17.0%) ชี้ให้เห็นจากกราฟความถี่การโพสต์รายชั่วโมง (Bluesky Posts Distribution by Hour of Day)
### **So what:**
คอนเทนต์หรือแคมเปญการตลาดที่เผยแพร่นอกช่วงเวลา 14:00–16:00 น. จะมีโอกาสเข้าถึงกลุ่มเป้าหมายไม่ถึง 1.2% ทำให้งบประมาณสื่อสื่อสารการตลาดมีประสิทธิภาพต่ำลง
### **Action:**
ทีมการตลาดกำหนดเวลาบรอดแคสต์และยิงโฆษณาหลักให้อยู่ในช่วง 14:00–15:30 น. แล้ววัดผลสำเร็จจากสัดส่วน Engagement Rate ที่คาดว่าจะเพิ่มขึ้นอย่างชัดเจนภายใน 1 เดือน
### **ความมั่นใจ:**
ปานกลาง เนื่องจากคำนวณจากชุดข้อมูลขนาดใหญ่จำนวน 96,500 โพสต์ ซึ่งแสดงพฤติกรรมช่วงเวลาที่ชัดเจน
